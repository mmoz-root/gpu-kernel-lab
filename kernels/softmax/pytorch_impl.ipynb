```python
import torch
```

```python
def softmax_ref(x):
    
    max = torch.max(x, dim=1, keepdim=True).values
    numerator = torch.exp(x - max)
    denominator = torch.sum(numerator, dim=1, keepdim=True)

    return numerator / denominator
```

```python
x = torch.tensor([
    [1000.0, 1001.0, 1002.0],
    [0.0, 0.0, 0.0]
])


result = softmax_ref(x)
expected = torch.softmax(x, dim=1)

print(result)
print(result.sum(dim=1))
print(torch.allclose(result, expected))
```
