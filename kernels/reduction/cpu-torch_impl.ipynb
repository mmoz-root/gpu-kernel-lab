```python
import torch

def check_input(x):
    assert x.device.type == "cpu"
    assert x.dtype == torch.float32
    assert x.ndim == 1
    assert x.is_contiguous()
    assert x.numel() > 0
```

```python
def custom_sum(x):
    check_input(x)

    result = 0.0

    for value in x:
        result += value.item()

    return result
```

```python
def custom_max(x):
    check_input(x)

    result = x[0].item()

    for value in x[1:]:
        candidate = value.item()

        if candidate > result:
            result = candidate

    return result
```

```python
def torch_sum(x):
    check_input(x)
    return torch.sum(x)

def torch_max(x):
    check_input(x)
    return torch.max(x)
```

```python
x = torch.tensor(
    [3.0, -2.0, 7.0, 1.5], dtype=torch.float32
)

manual_sum = custom_sum(x)
torchSum = torch_sum(x)

manual_max = custom_max(x)
torchMax = torch_max(x)

print("CPU sum:    ", manual_sum)
print("PyTorch sum:", torchSum.item())

print("CPU max:    ", manual_max)
print("PyTorch max:", torchMax.item())
```

```python
assert torch.isclose(
    torch.tensor(manual_sum), torchSum
)

assert manual_max == torchMax.item()
```
